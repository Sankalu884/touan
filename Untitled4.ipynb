```python
import csv
import re
from bs4 import BeautifulSoup
from google.colab import files

HTML_FILE = "短答式過去問題演習トレーニング刑法R7解答つき.html"
CSV_FILE = "anki_import_with_answers.csv"

def extract_questions():
    with open(HTML_FILE, "r", encoding="utf-8") as f:
        soup = BeautifulSoup(f, "html.parser")

    questions = []
    title_tds = soup.find_all("td", string=lambda t: t and "タイトル" in t)

    for title_td in title_tds:
        table = title_td.find_parent("table")
        if not table: continue

        # 執筆者名と引用元のdivを削除
        for div in table.find_all('div', style=True):
            if 'text-align' in div['style'].lower() and 'right' in div['style'].lower():
                div.extract()

        for div in table.find_all('div', class_=True):
            classes = div.get('class', [])
            if 'text-xs-start' in classes and 'text-sm-end' in classes:
                div.extract()

        title = title_td.find_next_sibling("td").get_text(strip=True) if title_td.find_next_sibling("td") else ""

        mondai_html = ""
        anschoi_options = []
        subques_options = []

        mondai_td = table.find("td", class_=lambda c: c and "dtlview-header" in c, string=lambda t: t and t.strip() == "問題")
        if not mondai_td:
            mondai_td = table.find("td", string=lambda t: t and t.strip() == "問題")

        if mondai_td and mondai_td.find_next_sibling("td"):
            mondai_next = mondai_td.find_next_sibling("td")

            for child in mondai_next.contents:
                if child.name == "span" and child.get("class") == ["anschoi"]:
                    anschoi_options.append(child.get_text(strip=True))
                elif child.name == "span" and child.get("class") == ["subques"]:
                    subques_options.append(child.get_text(strip=True))
                elif child.name == "div" and child.get("class") and "Repopup_rule" in child.get("class"):
                    continue
                else:
                    mondai_html += str(child)

        # <br>タグの統一と、直後の空白（全角・半角）削除
        mondai_html = mondai_html.replace("<br/>", "<br>").strip()
        mondai_html = re.sub(r'(<br\s*/?>)[　 ]', r'\1', mondai_html)

        while mondai_html.endswith("<br>"):
            mondai_html = mondai_html[:-4].strip()

        # ▼ 修正ポイント：解答形式の厳密な判定 ▼
        is_multi_keyword = re.search(r'([2２二]個|すべて)[選えら][びべ]', mondai_html)
        is_combo_keyword = "組合せ" in mondai_html or "組み合わせ" in mondai_html

        # 正誤問題特有の「1を、2を」という指示があるかをチェック
        is_tf_keyword = re.search(r'正しい場合[に]?は[1１]を', mondai_html) or re.search(r'誤っている場合[に]?は[2２]を', mondai_html)

        if is_multi_keyword and not is_combo_keyword:
            format_type = "複数選択"
        elif is_tf_keyword:
            # 「1を、2を」という指示がある場合のみ「正誤」とする
            format_type = "正誤"
        else:
            # それ以外（組合せタグがある場合や、単なる1〜5の選択肢の場合）はすべて「択一」とする
            format_type = "択一"

        # 形式に応じて、Ankiのフィールドに入れる「選択肢」のリストを決定
        if len(anschoi_options) > 0:
            options = anschoi_options
        else:
            options = subques_options

        # 正解の抽出
        seikai = ""
        seikai_td = table.find("td", class_=lambda c: c and "dtlview-header" in c, string=lambda t: t and "正解" in t)
        if not seikai_td: seikai_td = table.find("td", string=lambda t: t and t.strip() == "正解")
        if seikai_td and seikai_td.find_next_sibling("td"):
            seikai_raw = seikai_td.find_next_sibling("td").get_text(strip=True)
            if format_type == "複数選択":
                numbers = re.findall(r'[1-8１-８]', seikai_raw)
                if numbers: seikai = ", ".join([n.translate(str.maketrans('１２３４５６７８', '12345678')) for n in numbers])
                else: seikai = seikai_raw
            elif format_type == "正誤":
                numbers = re.findall(r'[12１２]', seikai_raw)
                if numbers: seikai = ", ".join([n.translate(str.maketrans('１２', '12')) for n in numbers])
                else: seikai = seikai_raw
            else:
                match = re.search(r'[1-8１-８]', seikai_raw)
                if match: seikai = match.group(0).translate(str.maketrans('１２３４５６７８', '12345678'))
                else: seikai = seikai_raw

        # 解説の抽出（<br>直後のスペースも削除）
        kaisetsu = ""
        kaisetsu_td = table.find("td", class_=lambda c: c and "dtlview-header" in c, string=lambda t: t and "解説(参照)" in t)
        if not kaisetsu_td: kaisetsu_td = table.find("td", string=lambda t: t and (t.strip() == "解説" or t.strip() == "解説(参照)"))
        if kaisetsu_td and kaisetsu_td.find_next_sibling("td"):
            kaisetsu = "".join([str(c) for c in kaisetsu_td.find_next_sibling("td").contents]).strip()
            kaisetsu = re.sub(r'(<br\s*/?>)[　 ]', r'\1', kaisetsu)

        # 出典の抽出
        shuten = ""
        shuten_td = table.find("td", class_=lambda c: c and "dtlview-header" in c, string=lambda t: t and "出典" in t)
        if not shuten_td: shuten_td = table.find("td", string=lambda t: t and t.strip() == "出典")
        if shuten_td and shuten_td.find_next_sibling("td"):
            shuten = shuten_td.find_next_sibling("td").get_text(strip=True)

        row = {
            "タイトル": title, "問題": mondai_html, "解答形式": format_type,
            "正解": seikai, "解説": kaisetsu, "出典": shuten
        }
        for i in range(8):
            row[f"選択肢{i+1}"] = options[i] if i < len(options) else ""
        questions.append(row)

    return questions

def write_csv(questions):
    fieldnames = ["タイトル", "問題", "選択肢1", "選択肢2", "選択肢3", "選択肢4", "選択肢5", "選択肢6", "選択肢7", "選択肢8", "正解", "解説", "出典", "解答形式"]
    with open(CSV_FILE, "w", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(questions)

if __name__ == "__main__":
    questions = extract_questions()
    write_csv(questions)
    print(f"{len(questions)}件の問題をCSVに抽出しました。")
    files.download(CSV_FILE)
```
